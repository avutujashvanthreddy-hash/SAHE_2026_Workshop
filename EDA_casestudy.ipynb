```python
data['Age'] = data['Age'].fillna(data['Age'].mean())

data.isnull().sum()
```

```python
data.drop(columns=['Cabin'], inplace=True)
data.isnull().sum()
```

```python
data.head(10)
```

```python
data.columns
```

```python
pclass = pd.get_dummies(data['Pclass'],
                        drop_first=True,
                       dtype=int)
pclass
```

```python
gender = pd.get_dummies(data['Sex'],drop_first=True,
                          dtype=int)
gender
```

```python
embarked = pd.get_dummies(data['Embarked'],drop_first=True,
                          dtype=int)
embarked
```

```python
final_data = pd.concat([data,pclass,gender,embarked],
                       axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId', 'Survived', 'Pclass', 'Name', 'Sex', 'Age', 'SibSp',
       'Parch', 'Ticket', 'Fare', 'Embarked'],inplace=True)
final_data.head()
```
